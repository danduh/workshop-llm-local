## How JSON Chat Messages Become LLM Prompt Text

This notebook demonstrates how SDK/JSON chat messages are converted into a plain text template that is actually sent to an LLM.

LLMs only see text. Servers convert structured JSON into formatted text prompts before sending them to the model.

### Step 1 — Example JSON input (like an SDK sends)

In practice, your application sends structured messages like this:

````python
messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "What color is the sky?"}
]

messages
````

### Step 2 — Server converts JSON into template text

A server or SDK converts messages into a prompt template before sending it to the LLM.

Below we simulate that conversion.

````python
def build_prompt(messages):
    role_tokens = {
        "system": "<|system|>",
        "user": "<|user|>",
        "assistant": "<|assistant|>"
    }

    parts = []
    for msg in messages:
        token = role_tokens[msg["role"]]
        parts.append(f"{token}\n{msg['content']}")

    # Assistant token added so model continues from there
    parts.append("<|assistant|>\n")
    return "\n".join(parts)

prompt = build_prompt(messages)
print(prompt)
````

### Step 3 — This text is sent to the LLM

The model receives only plain text like this:

```
<|system|>
You are helpful.
<|user|>
What color is the sky?
<|assistant|>
```

Then the LLM predicts the continuation — the assistant's answer.

Everything else (JSON, SDKs, tools) exists outside the model.

## Key takeaway for workshops

> Chat systems are just structured text converted into prompts.
> LLMs only perform next-token prediction on text.
