```python
from mlx_lm import load, generate

model, tokenizer = load("mlx-community/Phi-3-mini-4k-instruct-4bit")


prompt = "<|user|>\nWhat sky color at night?\n<|assistant|>\n"
out = generate(model, tokenizer, prompt, max_tokens=120)
print(out)
```
